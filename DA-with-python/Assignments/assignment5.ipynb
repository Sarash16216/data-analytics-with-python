QUESTION 1: Working with Lists
Problem Statement:

Create a list called numbers containing at least 8 integers.
Using a for loop, do the following:   

Print all numbers in the list.   

Print the sum of all numbers.   

Print only the even numbers.   

Requirements: Use at least one loop, use conditional statements, and display clear output.

```python
# Solution 1: Working with Lists
numbers = [12, 7, 15, 22, 8, 33, 40, 19]

print("All numbers in the list:")
total_sum = 0
even_numbers = []

for num in numbers:
    print(num, end=" ")
    total_sum += num
    if num % 2 == 0:
        even_numbers.append(num)

print(f"\nSum of all numbers: {total_sum}")
print(f"Even numbers: {even_numbers}\n")
```

reate a dictionary called student with the following keys:   name   age   course   marks (must store a list of 3 marks)   Using loops:   Print all keys and their values.   Calculate the average of the marks.   Display whether the student Passed (average $\ge 50$) or Failed.   Requirements: Dictionary must contain a list, use loops to calculate the average.

```python
# Solution 2: Dictionary Basics
student = {
    "name": "Aashir",
    "age": 12,
    "course": "Python Programming",
    "marks": [85, 90, 78]
}

print("Student Details:")
for key, value in student.items():
    print(f"- {key}: {value}")

total_marks = 0
for mark in student["marks"]:
    total_marks += mark

average_marks = total_marks / len(student["marks"])
print(f"Average Marks: {average_marks:.2f}")

if average_marks >= 50:
    print("Status: Passed\n")
else:
    print("Status: Failed\n")
```

Create a list called employees. Each employee must be stored as a dictionary containing:   

name   

department   

salary   

Using loops:   

Print details of all employees.   

Find the employee with the highest salary.   

Calculate the total salary expense.   

Requirements: Minimum 3 employees, use comparison logic, use loops to process data.

```python
# Solution 3: List of Dictionaries
employees = [
    {"name": "Alice", "department": "HR", "salary": 45000},
    {"name": "Bob", "department": "Engineering", "salary": 75000},
    {"name": "Charlie", "department": "Marketing", "salary": 60000}
]

print("Employee Details:")
total_salary_expense = 0
highest_salary = 0
top_employee = ""

for emp in employees:
    print(f"Name: {emp['name']}, Dept: {emp['department']}, Salary: ${emp['salary']}")
    total_salary_expense += emp['salary']
    
    if emp['salary'] > highest_salary:
        highest_salary = emp['salary']
        top_employee = emp['name']

print(f"Employee with Highest Salary: {top_employee} (${highest_salary})")
print(f"Total Salary Expense: ${total_salary_expense}\n")
```

Ask the user to enter numbers. Store the numbers in a list. Stop when the user enters -1.
After the loop ends:

Print the complete list.

Print the largest number.

Print the smallest number.

Requirements: Must use a while loop, store input in a list, handle user input correctly.

```python
# Solution 4: While Loop with User Input
user_numbers = []
print("Enter numbers to add to the list (Enter -1 to stop):")

while True:
    try:
        val = int(input("Enter number (-1 to quit): "))
        if val == -1:
            break
        user_numbers.append(val)
    except ValueError:
        print("Invalid input. Please enter a valid integer.")

print(f"Complete List: {user_numbers}")
if len(user_numbers) > 0:
    print(f"Largest Number: {max(user_numbers)}")
    print(f"Smallest Number: {min(user_numbers)}\n")
else:
    print("No numbers were entered.\n")
```

Ask the user to enter a sentence. Convert the sentence into a list of words. Use a dictionary to count how many times each word appears. Print the word frequency clearly.

Example Format:
apple: 3
banana: 2
orange: 1

Requirements: Use string splitting, use a dictionary for counting, use loops for processing.

```python
# Solution 5: Word Frequency Counter
sentence = input("Enter a sentence: ")

words_list = sentence.split()
word_counts = {}

for word in words_list:
    word_counts[word] = word_counts.get(word, 0) + 1

print("Word Frequency:")
for word, count in word_counts.items():
    print(f"{word}: {count}")
print()
```

Using nested loops, print a multiplication table from 1 to 5.   

Expected Output Format:
1 2 3 4 5
2 4 6 8 10
3 6 9 12 15
4 8 12 16 20
5 10 15 20 25   

Requirements: Must use nested loops, proper formatting required.

```python
# Solution 6: Nested Loops – Multiplication Table
print("Multiplication Table from 1 to 5:")

for i in range(1, 6):
    for j in range(1, 6):
        print(f"{i * j:<4}", end="")
    print() # Newline after each row
```
